**INNAPPROPRIATE DATA CLEANING STARTS**

```python
import numpy as np
```

```python
df.loc[df['product_price'] < 0, 'product_price'] = np.nan

df.loc[df['discount_percent'] < 0, 'discount_percent'] = np.nan

df.loc[df['past_return_rate'] < 0, 'past_return_rate'] = np.nan

df.loc[df['session_length_minutes'] < 0, 'session_length_minutes'] = np.nan

df.loc[df['num_product_views'] < 0, 'num_product_views'] = np.nan

df.loc[df['product_rating'] > 5, 'product_rating'] = np.nan
```

```python
df
```

```python
df.isnull().sum()
```

```python
df[['product_price',
    'discount_percent',
    'product_rating',
    'past_return_rate',
    'session_length_minutes',
    'num_product_views']].median()
```

```python
df['product_price'].fillna(df['product_price'].median(), inplace=True)

df['discount_percent'].fillna(df['discount_percent'].median(), inplace=True)

df['product_rating'].fillna(df['product_rating'].median(), inplace=True)

df['past_return_rate'].fillna(df['past_return_rate'].median(), inplace=True)

df['session_length_minutes'].fillna(df['session_length_minutes'].median(), inplace=True)

df['num_product_views'].fillna(df['num_product_views'].median(), inplace=True)
```

```python
df.isnull().sum().sum()
```

```python
cols = [
    'product_price',
    'discount_percent',
    'product_rating',
    'past_return_rate',
    'session_length_minutes',
    'num_product_views'
]

for col in cols:
    df[col] = df[col].fillna(df[col].median())
```
